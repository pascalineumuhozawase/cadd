```python
#Cell 1 — Imports & config

import os, json, shutil
from pathlib import Path
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.metrics import classification_report, confusion_matrix

CSV_PATH        = Path("../dataset/train.csv")
RAW_IMAGES_DIR  = Path("../dataset/train_images")
PROCESSED_DIR   = Path("../dataset/processed/train")
MODEL_OUT_DIR   = Path("../models")
MODEL_OUT_PATH  = MODEL_OUT_DIR / "mobilenetv2_cassava.keras"

IMG_SIZE        = (224, 224)
BATCH_SIZE      = 32
SEED            = 42
EPOCHS_HEAD     = 8
EPOCHS_FINETUNE = 5

CLASS_NAMES = ["CBB", "CBSD", "CGM", "CMD", "healthy"]
LABEL_MAP   = {0:"CBB", 1:"CBSD", 2:"CGM", 3:"CMD", 4:"healthy"}

print("TF:", tf.__version__, "| GPU:", tf.config.list_physical_devices("GPU"))
```

```python
#Cell 2 — Prepare dataset
if not (PROCESSED_DIR.exists() and any(PROCESSED_DIR.iterdir())):
    df = pd.read_csv(CSV_PATH)
    for name in CLASS_NAMES:
        (PROCESSED_DIR / name).mkdir(parents=True, exist_ok=True)
    n = 0
    for _, row in df.iterrows():
        src = RAW_IMAGES_DIR / row["image_id"]
        if src.exists():
            dst = PROCESSED_DIR / LABEL_MAP[int(row["label"])] / row["image_id"]
            if not dst.exists():
                shutil.copy(src, dst); n += 1
    print("Copied:", n)
else:
    print("Processed folder ready.")
```
